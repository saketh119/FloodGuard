# 01 Data Exploration

Load each dataset, inspect its structure, and record baseline observations before any cleaning.

```python
import pandas as pd

flood = pd.read_csv("../../../data/raw/ml/flood_risk_india.csv")
rainfall = pd.read_csv("../../../data/raw/ml/indian_rainfall_daily_measurements.csv")
events = pd.read_csv("../../../data/raw/ml/indofloods_flood_events.csv")
precipitation = pd.read_csv("../../../data/raw/ml/indofloods_precipitation_variables.csv")
catchment = pd.read_csv("../../../data/raw/ml/indofloods_catchment_characteristics.csv")

datasets = {
    "flood": flood,
    "rainfall": rainfall,
    "events": events,
    "precipitation": precipitation,
    "catchment": catchment,
}

datasets.keys()
```

## Dataset Inspection

Run the same inspection checks for every dataset without removing or altering any rows yet.

```python
for name, df in datasets.items():
    print(f"===== {name.upper()} =====")
    print(df.shape)
    display(df.head())
    df.info()
    display(df.describe(include="all"))
    print(df.columns.tolist())
    print(df.isnull().sum())
    print(df.duplicated().sum())
    print()
```

## Dataset Visualizations

Create a small set of plots for each dataset so you can spot missingness, distributions, and relationships before preprocessing.

```python
import matplotlib.pyplot as plt
import missingno as msn
import seaborn as sns


def visualize_dataset(name, df):
    print(f"===== {name.upper()} VISUALS =====")

    print("Missing values")
    msn.matrix(df)
    plt.show()

    numeric_df = df.select_dtypes(include="number")
    if not numeric_df.empty:
        print("Distribution")
        numeric_df.hist(figsize=(15, 10))
        plt.suptitle(f"{name} - numeric distributions")
        plt.show()
    else:
        print("Distribution skipped: no numeric columns found.")

    if numeric_df.shape[1] > 1:
        print("Correlation")
        plt.figure(figsize=(12, 8))
        sns.heatmap(numeric_df.corr(), annot=False, cmap="viridis")
        plt.title(f"{name} - correlation heatmap")
        plt.show()
    else:
        print("Correlation skipped: not enough numeric columns.")


for name, df in datasets.items():
    visualize_dataset(name, df)
    print()
```
